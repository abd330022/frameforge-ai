```python
import torch, os, time
print('PyTorch:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available(): print('GPU:', torch.cuda.get_device_name(0))
from silma_tts.api import SilmaTTS
tts = SilmaTTS()
reference_audio_file = '/content/silma-tts/src/silma_tts/infer/ref_audio_samples/ar.ref.24k.wav'
assert os.path.exists(reference_audio_file), f'Reference audio not found: {reference_audio_file}'
reference_audio_text = 'ويدقق النظر في القرآن الكريم وسائر الكتب السماوية ويتبع مسالك الرسل العظام عليهم الصلاة والسلام.'
print('Model loaded and Arabic reference audio found.')
```

## 1) عينة عربية فصحى
يستخدم هذا الاختبار عينة المرجع العربية المرفقة في مستودع النموذج. لا تستخدم صوت شخص حقيقي آخر دون موافقته.

```python
arabic_text = 'مرحباً بكم في مشروع فريم فورج. نحول الأفكار والقصص إلى مشاهد تنبض بالحياة.'
start = time.time()
wav_ar, sr_ar, spec_ar = tts.infer(ref_file=reference_audio_file, ref_text=reference_audio_text, gen_text=arabic_text, file_wave='/content/frameforge_silma_arabic.wav', seed=42, speed=1)
print(f'Arabic generation time: {time.time()-start:.2f}s | sample rate: {sr_ar}')
from IPython.display import Audio, display
display(Audio('/content/frameforge_silma_arabic.wav'))
```

## 2) عينة إنجليزية

```python
english_text = 'Welcome to FrameForge. Turn your ideas and stories into vivid animated scenes.'
start = time.time()
wav_en, sr_en, spec_en = tts.infer(ref_file=reference_audio_file, ref_text=reference_audio_text, gen_text=english_text, file_wave='/content/frameforge_silma_english.wav', seed=42, speed=1)
print(f'English generation time: {time.time()-start:.2f}s | sample rate: {sr_en}')
display(Audio('/content/frameforge_silma_english.wav'))
```
